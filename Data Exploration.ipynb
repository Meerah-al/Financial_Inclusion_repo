```python
data1['relationship_with_head'].value_counts().to_frame('Count').assign(Pecentage=lambda x: (x['Count'] / x['Count'].sum()*100).round(2))
```

```python
#Converting categorical to numerical so we can use knn
data1['relationship_with_head']=data1['relationship_with_head'].map({'Head of Household':0,'Spouse':1, 'Child':2, 'Parent':3, 'Other relative':4, 'Other non-relatives':5})
data1.head()
```

```python
#Choosing the features as x and the target as y
x=data1[['country','cellphone_access','location_type','relationship_with_head','gender_of_respondent', 'marital_status','education_level','job_type']]
y=data1['bank_account']
```

```python
#split our data with test size of 20%
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
knn=KNeighborsClassifier(n_neighbors=1) #build our knn classifier
```

```python
knn.fit(x_train,y_train) #Training KNN classifier
```

```python
y_pred = knn.predict(x_test)  #Testing
y_pred
```

```python
print('Acuuracy=',accuracy_score(y_pred,y_test))
```

```python
n_neighbors=30
scores=[]
for k in range(1,30):
    knn=KNeighborsClassifier(n_neighbors-k)
    knn.fit(x_train,y_train)
    y_pred=knn.predict(x_test)
    print('Accuracy for k=',k,'is:',round(accuracy_score(y_pred,y_test),2))
    scores.append(round(accuracy_score(y_pred,y_test),2))
```
